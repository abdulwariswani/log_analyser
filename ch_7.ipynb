```python
def isPhoneNumber(text):
    if len(text) != 12:
        return False
    for i in range(0, 3):
        if not text[i].isdecimal():
            return False
    if text[3] != '-':
        return False
    for i in range(4, 7):
        if not text[i].isdecimal():
            return False
    if text[7] != '-':
        return False
    for i in range(8, 12):
        if not text[i].isdecimal():
            return False
    return True

print('415-555-4242 is a phone number:')
print(isPhoneNumber('415-555-4242'))
print('Moshi moshi is a phone number:')
print(isPhoneNumber('Moshi moshi'))




message = 'Call me at 415-555-1011 tomorrow . 415-555-1011 is my office'
for i in range(len(message)):
    chunk = message [i:i+12]
    if isPhoneNumber (chunk):
        print('Phone number is found:'+ chunk)
print('Done')
```

```python
import re 
PhoneNumregex = re.compile(r'(\d\d\d)?-(\d\d\d-\d\d\d)') # d means numeral [0 to 9]
mo = PhoneNumregex.search('Call me at 415-555-1011 tomorrow . 415-555-1011 is my office')
mo2 = PhoneNumregex.search('fuckers number is 222-444-5555')
print(mo2.group(1))
print(mo2.group(2))
print(mo.group())
```

```python
# grouping with parantheses in regex
PhoneNumregex = re.compile(r'(\(\d\d\d\)) (\d\d\d-\d\d\d)')
mo = PhoneNumregex.search('My phone number is : (415) 555-1011') # if the phone no contain paranthesis
print(mo.group(1))
print(mo.group(2))
print(mo.group())
print(mo.groups())# to retrieve all the groups once 
areacode , mainnumber = mo.groups()
print(areacode,mainnumber)
```

```python
# matching multiple groups with pipe

heroregex = re.compile(r'Batman|tina fey')
m01 = heroregex.search('tina fey is Batman in disguise')
print(m01.group())

# can use pipe to match on of serveral pattern e.g
batRegex = re.compile(r'Bat(man|bat|copter|mobile)')
mo = batRegex.search('Batmobile lost a wheel')
print(mo.group())
print(mo.group(1))
```

```python

# short hand character classes
xmasregex=re.compile(r'\d+\s\w+')
print(xmasregex.findall('12 drummers, 11 pipers, 10 lords, 9 ladies, 8 maids, 7 swans, 6 geese, 5 rings, 4 birds, 3 hens, 2 doves, 1 partridge'))




# if it is too broad you can create your own classes 
vowelregex = re.compile(r'[aieouAEIOU]')
constantregex = re.compile(r'[^aieouAEIOU]') # inside square bracket their is no need to use \ for escape
print(constantregex.findall('Robocap eats baby food. BABY FOOD'))
print(vowelregex.findall('Robocap eats baby food. BABY FOOD'))
```

```python
# The caret and dollar sign characters
beginwithHello = re.compile(r'^Hello') # matched if begining of the sentence is Hello
print(beginwithHello.search('Hello world!'))
print(None == beginwithHello.search("i and the myself"))



# dollar sign to match the end of the string regex
endwithNumber=re.compile(r'\d$')
print(endwithNumber.search("your number is :45"))
endwithNumber.search("your number is forty five") == None # put d+ for one or more numeric digit at the end


```

```python
# the wildcard Character to match character except newline 
atRegex = re.compile(r'.at')
atRegex.findall('The cat in the hat sat on the flat mat') # match only one character before the regex
```

```python
#Matching everything with dot-star .It uses the greedy mode so use (.*)
nameRegex = re.compile(r'First Name:(.*) Last Name:(.*)')
mo = nameRegex.search('First Name:AL Last Name:flick')
print(mo.group(1))
print(mo.group(2))
# to match in a non greedy-way
nongreedyRegex = re.compile(r'<.*?>')
mo = nongreedyRegex.search('<to serve man> for dinner>')
print(mo.group())
#greedy version
greedyversion = re.compile('<.*>')
mo1 = greedyversion.search('<to serve man> for dinner>')
mo1.group()
```

```python
# matching newline with Dot character
noNewLineregex = re.compile('.*')
print(noNewLineregex.search('Serve the public trust. \n Protect the innocent.\n Uphold the law.').group())

newLineRegex = re.compile('.*',re.DOTALL)
newLineRegex.search('Serve the public trust. \n Prottect the innocen.\n Uphold the law').group()

```

```python
# Making regex case-insensitive 
robocop = re.compile(r'robocop',re.I)
print(robocop.search('RoboCop is part machine, part human ,all cop').group())
print(robocop.search('ROBOCOP protects the innocent.').group())
```
