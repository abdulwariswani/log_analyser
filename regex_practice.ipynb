```python
# Day_3
# Validate a date in YYYY-MM-DD format 
import re
infor = input("Enter the date: ")
def check(infor):
    pattern = r'\d{4}-(0[0-9]|1[0-2])-(0[0-9]|[12][0-9]|3[01])'
    match = re.search(pattern , infor)
    if match:
        print('Valid date format',match.group())
    else:
        print('Enter a valid date format.')
check(infor)


#Find all words that contain three consecutive numbers

infor_1 = input("Enter the word: ")
def word(infor_1):
    pattern_1 = r'(\b\w*([a-zA-Z])\2{2,}\w*\b)'
    match_1 = re.search(pattern_1,infor_1)
    if match_1:
        print(match_1.group(1))
        print('\n')
        print(match_1.group(2))
    else:
        print('there is no word of the requirement.')

word(infor_1)

#check if the string is single phone number 
def Ph_no():
    pattern = r'^\+\d{2}-\d{10}$'
    print(re.search(pattern,'+91-9876543210'))
    return None

Ph_no()
```

```python
#DAY_4
# extract all words that doesnt contain any vowels
import re
def NoVowels():
    pattern = r'\b[^aeiouAEIOU\s]+\b'
    sh = re.findall(pattern,'fly by rhythm of the world')
    print(sh)
    return None 
NoVowels()
# check if string is a valid hex color code
def hexColor():
    pattern = r'#[A-F0-9]{6}'
    to = re.findall(pattern ,' #FFA500 , #ffa50')
    print(to)
    return None 
hexColor()

# validate the string contain only letters and digits and spaces
def string():
    pattern_1= r'^[A-Za-z0-9\s]+$'
    if re.fullmatch(pattern_1,'Hello 123 fuck the world 77 '):
        return True
    else:
        return False
string()
```

```python
# DAY-5
# extract all email-addresses in the text
import re
def extract():
    pattern =r'\b[A-Za-z0-9.]+@[a-zA-Z0-9]+\.[A-Za-z]{2,4}\b'
    text = "Contact us: help@site.com, info@my.org"
    yes = re.findall(pattern,text)
    print(yes)
    return None 

extract()

#Swap parts 
def swap():
    pattern = r'([a-z]+),\s+([a-z]+)'
    text = "Doe, John"
    result = re.sub(pattern ,r'\2 \1',text,flags=re.I)
    print(result)
    return None 

swap()

# find first repeated word in row 

def repeated():
    pattern = r'([a-zA-Z]+)\s\1'
    text = "the the cat sat"
    yes = re.search(pattern,text)
    print(yes.group())
    return None 
repeated()
```

```python
# DAY6
# replace every digit in string with *
def Replace():
    pattern = r'\d'
    text = 'My pin is 1234'
    new = re.sub(pattern,'*',text)
    print(new)
    return None 
Replace()

# split sentence 
def SplitChar():
    pattern = r'([a-zA-Z]+)[ \t\n]?'
    text = "Hi there\tworld"
    yes = re.findall(pattern,text)
    print(yes)
    return None 

SplitChar()

# Replace multiple spaces with a single space
def SingleSpace():
    pattern_1 = r'^\s+|\s+$'
    pattern = r'\s+'
    text = " too many spaces "
    new = re.sub(pattern,' ',text)
    new_1 = re.sub(pattern_1,'',new)
    print(new_1)
    return None 
SingleSpace()
```

```python
# DAY-7
# check if the number is optional comma,(commas every 3 digit from the right)
import re 
def comma():
    pattern =r'^\d{1,3}(?:,\d{3})*$'
    text = "12,345,343,234"
    yes = re.findall(pattern,text)
    print(yes)

comma()
# validate password 

def passn():
    pattern = r'^(?=.*[A-Z])(?=.*[a-z])(?=.*\d).{8,}$'
    text = "p@sSw0rd"
    print(re.search(pattern,text).group())
passn()

def extract():
    pattern = r'(?<=@)[^<]+'
    text = "user@example.com"
    yes = re.search(pattern,text)
    print(yes.group())

extract()

```

```python
# DAY_8
# extract the protocol , domain and path from the URL
import re
def parsing():
    pattern = r'(https|http)://(www\.)?([A-Za-z.]+)([/A-Z0-9a-z]+)'
    text = 'https://www.example.com/page'
    matches  = re.findall(pattern,text)
    return matches
print(parsing())

def Log():
    pattern = r'ERROR:\s.+'
    text = '[2026-05-02 10:30:45] ERROR: Something failed'
    yes = re.search(pattern,text).group()
    return yes
print(Log())
```

```python
#Day-9
# Validate IPV4 address 0-255 separated by dots and no leading zeros 
import re
def IP_v():
    patteren =re.compile(r'''([0-9]|[1-9][0-9]|1[0-9]{2}|2[0-4][0-9]|25[0-5])\.
    ([0-9]|[1-9][0-9]|1[0-9]{2}|2[0-4][0-9]|25[0-5])\.
    ([0-9]|[1-9][0-9]|1[0-9]{2}|2[0-4][0-9]|25[0-5])\.
    ([0-9]|[1-9][0-9]|1[0-9]{2}|2[0-4][0-9]|25[0-5])
    ''',re.VERBOSE)
    text = "192.168.1.1"
    no = re.search(patteren,text).group()
    print(no)

IP_v()

# file extension --> extract .pdf and .jpj
def extract():
    patteren = r'.*\.pdf|.*\.jpg'
    text_files = ['report.pdf', 'image.jpg']
    yesp = [re.findall(patteren,text) for text in text_files ]
    print(yesp)

extract()
```
