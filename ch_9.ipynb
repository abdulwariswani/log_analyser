# WALKING A TREE DIRECTORY
# learn to os.walk(). Very usefull 



```python
import os

# Ensure you use different names for the list and the individual item
for foldername, subfolders, filenames in os.walk('walk'):
    print('The current folder is ' + foldername)
    
    # Use 'subfolder' (singular) to loop through 'subfolders' (plural)
    for subfolder in subfolders:
        print('SUBFOLDER OF ' + foldername + ': ' + subfolder)
    
    # Use 'filename' (singular) to loop through 'filenames' (plural)
    for filename in filenames:
        print('FILE INSIDE ' + foldername + ': ' + filename)
        
    print('') # Prints a newline after each folder's contents
```

# to zip file using python module zipfile

```python
import zipfile, os 
os.chdir(os.getcwd())
yes = zipfile.ZipFile('file.zip')
yes.namelist() # will show all the files folders and subfolders 
spaminfo = yes.getinfo('new.txt')
spaminfo.file_size
spaminfo.compress_size
f'compressed file is {round(spaminfo.file_size/spaminfo.compress_size,2)}x smaler'





# extracting zip files
yes.extractall()
# if want to extract a single file from the zip 
yes.extract('filename','destination_path')
#filename should be among the namelist()
```

```python
# CREATING AND ADDING TO THE ZIP FILES
import zipfile
newZip = zipfile.ZipFile('new.zip','w')
newZip.write('spam.txt',compress_type =zipfile.ZIP_DEFLATED)
newZip.close()
```
