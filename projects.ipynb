```python
#mcb.pyw - saves and loads piece of text to the clipboard 
# Usage: py.exe mcb.py save <keyword> - Saves clipboard to keyword.
# py.exe mcb.py <keyword> - Loads keyword to clipboard.
# py.exe mcb.py list - Loads all keywords to clipboard.

import shelve ,pyperclip, sys
mcbShelf = shelve.open('mcb')
# save clipboard content 

if len(sys.argv) == 3 and sys.argv[1].lower() == 'save':
    mcbShelf[sys.argv[2]] = pyperclip.paste()

elif len(sys.argv) == 3 and sys.argv[1].lower() == 'delete':
    if sys.argv[2] in mcbShelf:
        del mcbShelf[sys.argv[2]]
    else:
        print("key doesn't exist")
elif len(sys.argv) == 2 and sys.argv[1].lower() =='del': # can use loop to del manually works everwhere, instead of using clear()
    mcbShelf.clear()

elif len(sys.argv) == 2:
    # list keywords and load content
    if sys.argv[1].lower() == 'list':
        pyperclip.copy(str(list(mcbShelf.keys())))
    elif sys.argv[1] in mcbShelf:
        pyperclip.copy(mcbShelf[sys.argv[1]]) 

   
mcbShelf.close()







# Step 1: copy some text manually
# e.g. copy "hello world"
#python mcb.py save hello
#python mcb.py list
# paste → should show ['hello']
#python mcb.py hello
```

```python
import re, pyperclip

Copied_text = pyperclip.paste()

Phoneregex = re.compile(r'''
                        (\d{3}|\(\d{3}\))?            # Area code (optional)
                        (\s|-|\.)?                    # Separator
                        (\d{3})                       # First 3 digits
                        (\s|-|\.)                   # Separator
                        (\d{4})                       # Last 4 digits
                        (\s*(ext|x|ext)\s*\d{2,5})?   # Extension (optional and greedy)
                        ''', re.VERBOSE)

Emailregex = re.compile(r'''(
                        [a-zA-z0-9._%+-]+
                        @
                        [a-zA-Z0-9._%-]+
                        (\.[a-zA-Z]{2,4})
                        )''', re.VERBOSE)
matches = []
for groups in Emailregex.findall(Copied_text):
    matches.append(groups[0])

for groups in Phoneregex.findall(Copied_text):
    # Join the groups together to recreate the full string
    phone_num = '-'.join([groups[0], groups[2], groups[4]])
    if groups[5] != '':
        phone_num += ' ' + groups[5]
    matches.append(phone_num)

# Copy the cleaned results back to clipboard



if len(matches) > 0 :
    pyperclip.copy('\n'.join(matches))
    print("copied to clipboard.\n")
    print('\n'.join(matches))
else:
    print("No phone number or email found ")
```

```python
# strong password detection. 
import re

infor = input("Enter the password: \n")
def secure_pass(infor):
    strong_passwd = re.compile(r'''
     (?=.*[a-z])
     (?=.*[A-Z])
     (?=.*[0-9])
     (.{8,}$)                           
     ''',re.VERBOSE)
    match = strong_passwd.search(infor)
    if match:
        result = match.group()
        print(f'The {result} is secure')
    else:
        print("The given password is not secure")
    return None 


secure_pass(infor)


# replicate strip() function in python. 

#word = input('Enter the word: \n')
# char = input('Enter word you want remove')

def regex_strip(word,char=None):
    if char is None:
        seq = r'(\s+)|(\s+$)'
        return re.sub(seq,'',word)
        
      
    else:
        escaped = re.escape(char)
        pattern = f'^[{escaped}]+|[{escaped}]+$'
        return re.sub(pattern,'',word)
        
    
    
print(regex_strip("   hello \t "))         # "hello"
print(regex_strip("aabchelloabc", "abc"))  # "hello"
print(regex_strip("aabbcadaa", "ab"))      # "cad"
```

```python
import re, os , shutil
# to european DD-MM-YYYY
date_pattern = re.compile(r"""
                        ^(.*?)     #all text before date
                        ((0|1)?\d)-      # one or  two digits for the month
                        ((0|1|2|3)?\d)- # one or two digits for the day
                        ((19|20)\d\d) #four digits for the year
                        (.*?)$ # all text after the day                          
                          """,re.VERBOSE)
# loop over the files in the current working directory

for amerFilename in os.listdir('.'):
    mo = date_pattern.search(amerFilename)
    # skip file without name
    if mo == None :
        continue 
    #Get the different parts of the filename. 
    beforePart  = mo.group(1)
    monthPart = mo.group(2)
    dayPart = mo.group(4)
    yearPart = mo.group(6)
    afterPart = mo.group(8)
    # Form the european-style filename.
    euroFilename  = beforePart + dayPart + '-' + monthPart + '-' + yearPart + '-' + afterPart
    absWorkingDir = os.path.abspath('.')
    amerFilename = os.path.join(absWorkingDir,amerFilename)
    euroFilename = os.path.join(absWorkingDir,euroFilename)
    # Rename the files.
    print(f'Remaining {amerFilename} to {euroFilename}')
    shutil.move(amerFilename, euroFilename)



```

```python
#! python3
#Copies entire folder  and its contents into a Zip file whose name increments 
import zipfile , os
def backuptoZip(rename):
    # back up the entire content of 'folder' into a zipfile
    rename  = os.path.abspath(rename)
    number = 1
    while True:
        zipFilename = os.path.basename(rename) + '_' + str(number)+'.zip'
        if not os.path.exists(zipFilename):
            break
        number = number +1
        #Creating a ZIP file
        backupzip = zipfile.ZipFile('zipfilename','w')
        # Walk the entire folder  tree and compress the file in each folder
        for foldername, subfolders, filenames in os.walk(rename):
            print(f'Adding files in {foldername}')
            # add current folder to the zipfile 
            backupzip.write(foldername)
            # add all the files in the folder to the zip file.
            for filename in filenames:
                newBase =  os.path.basename(rename) + '_'
                if filename.startswith(newBase) and filename.endswith('.zip'):
                    continue # dont backup the backup ZIF files
            backupzip.write(os.path.join(foldername,filename))
    backupzip.close()
    print('Done.')

backuptoZip('/Users/abdulwariswani/Documents/WORKSPACE/plang')
```
